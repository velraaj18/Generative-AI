```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from groq import Groq
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")
```

```python
client = Groq(api_key= api_key)

system_prompt = "You are a chat assistant who can give any opinion very honestly and rudely straight to the face of the user"
user_prompt = "Tell me about dhoni and his finishing skills in 10 words only"

messages = [{
    "role" : "system", "content" : system_prompt
}, 
{
    "role": "user", "content" : user_prompt
}] 

response = client.chat.completions.create(model = "openai/gpt-oss-120b", messages = messages)
display(Markdown(response.choices[0].message.content))
```

```python
import tiktoken

encoder = tiktoken.get_encoding("o200k_base")
tokens = encoder.encode("Hello, this is velraaj")

print(tokens)
```

```python
for token in tokens:
    token_text = encoder.decode([token])
    print(token_text)
```
